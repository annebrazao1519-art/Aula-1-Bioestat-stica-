#2. Na população de insetos (10 min)
##Em uma população de abelhas, 30% dos indivíduos apresentam uma determinada característica genética.
##Um pesquisador captura aleatoriamente 10 abelhas.

##Questões
##a) Qual distribuição deve ser utilizada? Explique por quê.
###Resposta: Binomial, porque delimita o número de ensaios (n=10) e forneceu uma probabilidade conhecida (p=0.30)

##b) Qual é a probabilidade de exatamente 3 abelhas apresentarem a característica?
###Binomial, que apresenta a seguinte forma:
###$$P(X = k) = \binom{n}{k} \cdot p^k \cdot (1 - p)^{n - k}$$Substituindo os valores para $n = 10$, $k = 3$ e $p = 0{,}30$:
###Cálculo da combinação:$$\binom{10}{3} = \frac{10!}{3! \cdot (10 - 3)!} = \frac{10 \cdot 9 \cdot 8}{3 \cdot 2 \cdot 1} = 120$$
###Cálculo das potências:$$p^3 = 0{,}30^3 = 0{,}027$$$$(1 - p)^{10 - 3} = 0{,}70^7 \approx 0{,}0823543$$
###Cálculo da probabilidade final:$$P(X = 3) = 120 \cdot 0{,}027 \cdot 0{,}0823543 \approx 0{,}2668279$$

###A probabilidade de apresentar a característica é de aproximandamente 26,68%, ou seja, em um n amostral de 10 abelhas, cerca de 3 abelhas terão essa característica.

##c) Qual é a probabilidade de pelo menos 1 das 10 abelhas apresentar a característica?
###Como a questão pede que "pelo menos" 1 abelha apresente a característica, 2 podem possuir, 3 podem possuir... 10 podem possuir, ou seja um evento complementar. Ou seja:
###"Pelo menos 1" significa $P(X \ge 1)$, que é o mesmo que $1 - P(X = 0)$ (o complemento de nenhuma abelha ter a característica). Sendo assim a probabilidade de fracasso será o cálculo da probabilidade de nenhuma abelha haver essa característica.
###Calculando $P(X = 0)$ (nenhuma abelha com a característica):$$P(X = 0) = \binom{10}{0} \cdot (0{,}30)^0 \cdot (0{,}70)^{10}$$$$P(X = 0) = 1 \cdot 1 \cdot (0{,}70)^{10} \approx 0{,}0282475$$
###Calculando o complemento:$$P(X \ge 1) = 1 - P(X = 0)$$$$P(X \ge 1) = 1 - 0{,}0282475 = 0{,}9717525$$
###Então a probabilidade de pelo emnos 1 abelgha apresetar a característica específica é de aproximadamente 97,18%.


#3. Na observação ecológica (10 min)
##Durante levantamentos em uma área de cerrado, pesquisadores registram em média 2 encontros com uma determinada espécie de mamífero por hora.
##Questões
##a) Qual distribuição seria mais adequada para assumir nesse caso?
###Poisson, porque essa distribuição calcula a probabilidade de ocorrência de um evento aconter em uma quantidade X de vezes em um intervalo de tempo definido. Nesse caso, a observação ecológica calcula a média de encontra da espécie por hora, que se enquadra na fórmula descrita.

##b) Qual é a probabilidade de ocorrerem exatamente 3 encontros em uma hora?
###Utilizamos a fórmula de Poisson, em que $P(X = 3)$ para $\lambda = 2$
###$$P(X = 3) = \frac{e^{-2} \cdot 2^3}{3!}$$
###$2^3 = 8$
###$3! = 3 \times 2 \times 1 = 6$
###$e^{-2} \approx 0{,}135335$
###$$P(X = 3) = \frac{0{,}135335 \times 8}{6} = \frac{1{,}08268}{6} \approx 0{,}180447$$
###A probabilidade de ocorrer exatamente 3 encontros em uma hora é de aproximadamente 18,04%.

##c) Qual é a probabilidade de nenhum encontro ocorrer durante uma hora?
###Seguimos na mesma intenção do exercício anterior, onde $P(X = 0)$ para $\lambda = 2$
###$$P(X = 0) = \frac{e^{-2} \cdot 2^0}{0!}$$
###$2^0 = 1$
###$0! = 1$
###$$P(X = 0) = e^{-2} \approx 0{,}135335$$
###Então, a probabilidade de nenhum encontro ocorrer durante uma hora é de aproximadamente 13,53%

##d) Qual é a probabilidade de ocorrerem mais de 2 encontros em uma hora?
###Nesse caso queremos encontrar $P(X > 2)$, ou seja, a probabilidade de ocorrerem 3, 4, 5... até infinitos encontros. E como a soma de todas as probabilidades é $1$ ($100\%$), usamos o evento complementar:
###Já temos $P(X = 0) \approx 0{,}135335$. Vamos calcular $P(X = 1)$ e $P(X = 2)$:
###Para $k = 1$:$$P(X = 1) = \frac{e^{-2} \cdot 2^1}{1!} = \frac{0{,}135335 \times 2}{1} \approx 0{,}270670$$
###Para $k = 2$:$$P(X = 2) = \frac{e^{-2} \cdot 2^2}{2!} = \frac{0{,}135335 \times 4}{2} \approx 0{,}270670$$
###Somando $P(X \le 2)$:$$P(X \le 2) = 0{,}135335 + 0{,}270670 + 0{,}270670 = 0{,}676675$$
###Calculando $P(X > 2)$:$$P(X > 2) = 1 - 0{,}676675 = 0{,}323325$$$
###Ou seja, a probabilidade de ocorreram mais de 2 encontros em um ahora é de aproximadamente 32,33%

#4. Desafio final — Bernoulli versus Binomial versus Poisson? (5 min)
##Considere o seguinte experimento:
##Um ecólogo examina uma árvore e verifica se ela está infestada por uma determinada espécie de inseto.
##a) Qual é a distribuição se ele examinar uma única árvore?
### Bernoulli, pois a situação analisada de infestação ou não, ou sejam, rotulando como sucesso ou fracasso.

##b) Qual é a distribuição se ele examinar 50 árvores e quiser saber quantas estão infestadas?
###Binomial, pois se analisa o número de ensaios (n=50), que tem o resultado que está sendo ser observado, que no caso é a infestação,

##c) Qual é a principal diferença entre os dois modelos?
###A principal diferença é o número de ensaios
###Bernoulli: Considera uma única tentativa (n=1), resultando em um resultado simples de 0 (não infestada) ou 1 (infestada)
###Binomial: É a soma de n tentativas de Bernoulli independentes e idênticas, o resultado é a contagem de sucessos, podendo variar de 0 até n.

##Para discutir com a turma
##Complete a frase:
###Bernoulli é adequada quando realiuzamos um único ensaio (n=0), tendo apenas dois resultados possíveis (sucesso ou fracasso).
###Binomial é adequada quando realizamos um número fixo (n) de ensaios de Bernoulli, resultando na contagem de um número total de sucessos.
###Poisson é adequada quando contamos o número de ocorrência em um evento discreto em um intervalo contínuo (tempo, área ou volume), conhecendo apenas a taxa média de ocorrências ($\lambda$).
