```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df = pd.read_excel('/content/tarefa03.xlsx')
```

```python
df
```

```python
#frequência absoluta da variável nominal: tipo sanguíneo
freq_tipo_sanguíneo = df["tipo sanguíneo"].value_counts()
freq_tipo_sanguíneo
```

```python
#frequência relativa da variável nominal: tipo sanguíneo
freq_rel_tipo_sanguíneo= df["tipo sanguíneo"].value_counts(normalize=True)*100
freq_rel_tipo_sanguíneo
freq_rel_tipo_sanguíneo.round(1)
```

```python
#frequência absoluta da variável discreta
freq_doações_por_estudante= df['quantidade de doações'].value_counts (normalize=False)
freq_doações_por_estudante
```

```python
#frequência relativa da variável discreta: quantidade de doações
freq_doações_por_estudante= df["quantidade de doações"].value_counts(normalize=True).sort_index()*100
freq_doações_por_estudante.round(1)
```

#Gráficos

##Gráfico de barras

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x="tipo sanguíneo")
plt.title("Frequência de tipo sanguíneo")
plt.xlabel("tipo sanguíneo")
plt.ylabel("frequência")
plt.show()
```

##Histograma

```python
plt.figure(figsize=(6,4))
sns.histplot(data=df,
             x="quantidade de doações",
             bins=[0, 1, 2, 3, 4],
             kde=False)
plt.title("Frequência da quantidade de doações")
plt.xlabel("Quantidade de doações")
plt.ylabel("Frequência")

plt.xticks([0, 1, 2, 3, 4])

plt.show()
```

##Gráfico de dispersão

```python
plt.figure(figsize=(9,5))
sns.scatterplot(data=df,
                x="estudante",
                y="quantidade de doações",
                hue="tipo sanguíneo",
                style="tipo sanguíneo",
                s=120,
                palette="tab10")

plt.title("Quantidade de doações por estudante vs. tipo sanguíneo")
plt.xlabel("Estudante (ID)")
plt.ylabel("Quantidade de doações")

plt.xticks(range(1,16))
plt.yticks([0,1,2,3])
plt.grid(True, linestyle="--", alpha=0.4)
plt.legend(title="Tipo sanguíneo",
           bbox_to_anchor=(1,1),
           loc="upper left")
plt.show()
```
