```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df = pd.read_excel('/content/tarefa03.xlsx')
```

```python
df
```

```python
#frequência absoluta da variável nominal: tipo sanguíneo
freq_tipo_sanguíneo = df["tipo sanguíneo"].value_counts()
freq_tipo_sanguíneo
```

```python
#frequência relativa da variável nominal: tipo sanguíneo
freq_rel_tipo_sanguíneo= df["tipo sanguíneo"].value_counts(normalize=True)*100
freq_rel_tipo_sanguíneo
freq_rel_tipo_sanguíneo.round(1)
```

```python
#frequência absoluta da variável discreta
freq_doações_por_estudante= df['quantidade de doações'].value_counts (normalize=False)
freq_doações_por_estudante
```

```python
#frequência relativa da variável discreta: quantidade de doações
freq_doações_por_estudante= df["quantidade de doações"].value_counts(normalize=True).sort_index()*100
freq_doações_por_estudante.round(1)
```

#Gráfico de barras

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x="tipo sanguíneo")
plt.title("Frequência de tipo sanguíneo")
plt.xlabel("tipo sanguíneo")
plt.ylabel("frequência")
plt.show()
```
