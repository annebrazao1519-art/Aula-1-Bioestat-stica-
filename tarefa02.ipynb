```python
import pandas as pd
import matplotlib.pyplot as plt

# Ler a tabela
dados = pd.read_excel("dados_semana02.xlsx")

# Criar as faixas de idade
faixas = pd.cut(
    dados["Idade (anos)"],
    bins=[0, 8, 16, 25, float("inf")],
    labels=["0–8 anos", "9–16 anos", "17–25 anos", "26+ anos"]
)

# Contar quantas árvores existem em cada faixa
quantidade = faixas.value_counts().sort_index()

# Criar o gráfico
plt.figure(figsize=(9, 7))

fatias, textos, porcentagens = plt.pie(
    quantidade,
    autopct="%1.1f%%",
    startangle=90
)

# Criar os textos da legenda
legenda = [
    f"{faixa} — {qtd} árvores"
    for faixa, qtd in zip(quantidade.index, quantidade)
]

# Criar a legenda
plt.legend(
    fatias,
    legenda,
    title="Faixa de idade",
    loc="center left",
    bbox_to_anchor=(1, 0.5)
)

plt.title(
    "Distribuição das árvores por faixa etária",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    df['Tamanho (m)'],
    df['Idade (anos)'],
    color='forestgreen',
    edgecolor='black'
)

plt.title(
    'Relação entre Tamanho e Idade das Árvores',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Tamanho (m)', fontsize=12)
plt.ylabel('Idade (anos)', fontsize=12)

plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```
