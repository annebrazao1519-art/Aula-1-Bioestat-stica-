#**Exercício 2**
##a) H0 (Hipótese Nula): =2,0cm (o crescimento médio mensal das mudas é igual ao valor de referência de 2,0cm).
##H1 (Hipótese Alternativa): 2,0cm (o crescimento médio mensal das mudas é diferente de 2,0cm).


```python
# Parâmetro populacional sob H0
p_0 = 0.70
```

##b) É utilizado o método de Teste t de Student para uma Amostra. O objetivo é comparar a média de uma única amostra quantitativa com um valor populacional de referência ($\mu_0 = 2{,}0$), a variável apresenta distribuição aproximadamente normal e o desvio-padrão populacional ($\sigma$) é desconhecido, necessitando ser estimado pelo desvio-padrão amostral ($s$).



##c)A proporção amostral (p ou pest) é a razão entre o número de sementes germinadas (x=78) e o número total de sementes testadas (n=100):
$$\hat{p} = \frac{x}{n} = \frac{78}{100} = 0{,}78 \quad (78\%)$$


```python
x = 78 # Sementes que germinaram
n = 100
# Total de sementes plantadas
p_est = x / n
```

##d) Substituindo os valores conhecidos ($\hat{p} = 0{,}78$, $p_0 = 0{,}70$, $n = 100$):   

##Erro-padrão da proporção:
$$\sigma_{\hat{p}} = \sqrt{\frac{0{,}70 \times (1 - 0{,}70)}{100}} = \sqrt{\frac{0{,}21}{100}} = \sqrt{0{,}0021} \approx 0{,}04583$$
##Estatística $Z$:$$Z = \frac{0{,}78 - 0{,}70}{0{,}04583} = \frac{0{,}08}{0{,}04583} \approx 1{,}7457$$



```python
import numpy as np

p_0 = 0.70
n = 100
p_est = 78 / 100

# Cálculo do erro-padrão sob a hipótese nula
erro_padrao = np.sqrt((p_0 * (1 - p_0)) / n)

# Cálculo da estatística Z
z_calc = (p_est - p_0) / erro_padrao

print(f"Erro-padrão sob H0: {erro_padrao:.5f}")
print(f"Estatística de teste (Z): {z_calc:.4f}")
```

##e)Para um teste unilateral à direita com nível de significância $\alpha = 0{,}05$ ($5\%$), busca-se o valor na distribuição normal padrão que deixa $5\%$ da probabilidade na cauda superior ($Z_{0{,}95}$):   $Z_{\text{crítico}}$: $1{,}6449$

```python
import scipy.stats as stats

alpha = 0.05
# Quantil da distribuição normal padrão (área acumulada de 0.95)
z_critico = stats.norm.ppf(1 - alpha)

print(
    f"Valor crítico Z (alpha = {alpha}, unilateral à direita): {z_critico:.4f}"
)
```

##f)Decisão: Rejeita-se $H_0$.
##Justificativa: Como $Z_{\text{calculado}} = 1{,}7457$ é superior a $Z_{\text{crítico}} = 1{,}6449$, a estatística cai na região de rejeição da hipótese nula. Além disso, o p-valor associado é $p \approx 0{,}0404$, que é menor que $\alpha = 0{,}05$.


```python
import scipy.stats as stats

p_valor = 1 - stats.norm.cdf(z_calc)

print(f"Z calculado: {z_calc:.4f}")
print(f"Z crítico:   {z_critico:.4f}")
print(f"p-valor:     {p_valor:.4f}")

if z_calc > z_critico:
    decisao = "Rejeita-se H0"
else:
    decisao = "Não se rejeita H0"

print(f"Decisão estatística: {decisao}")
```

##g)Sim. Com a rejeição da hipótese nula ao nível de significância de $5\%$ ($Z = 1{,}7457$; $p = 0{,}0404$), há evidências estatísticas suficientes na amostra para sustentar a afirmação dos pesquisadores de que a taxa de germinação das sementes da espécie nativa sob condições controladas é superior a 70%.

##b) Devemos utilizar o Teste Qui-quadrado ($\chi^2$) para uma variância, pois o teste $t$ serve para avaliar a média populacional e não a dispersão ou variabilidade dos dados.



##c) Aplicando a fórmula com $n = 12$, $s^2 = 0{,}40$ e $\sigma_0^2 = 0{,}25$, a estatística qui-quadrado calculada resulta em $17{,}60$.

```python
# questão c:
n = 12
s2 = 0.40
sigma0_sq = 0.25

chi2_calc = ((n - 1) * s2) / sigma0_sq
print(f"Estatística Qui-quadrado calculada: {chi2_calc:.2f}")
```

##d) O número de graus de liberdade é igual a $11$, obtido pela subtração de uma unidade do tamanho total da amostra ($12 - 1$).

```python
# Questão d:
n = 12
df = n - 1
print(f"Graus de liberdade (df): {df}")
```

## e) O valor crítico da distribuição qui-quadrado para um teste unilateral à direita, com nível de significância de $5\%$ ($\alpha = 0{,}05$) e $11$ graus de liberdade, é de aproximadamente $19{,}68$.Python

```python
# Questão e:
import scipy.stats as stats

df = 11
alpha = 0.05
chi2_crit = stats.chi2.ppf(1 - alpha, df=df)
print(f"Valor crítico (alpha = 0.05, df = 11): {chi2_crit:.4f}")
```

##f) Como a estatística calculada ($17{,}60$) é menor do que o valor crítico ($19{,}68$), ela não se encontra na região de rejeição e a decisão é de não rejeitar $H_0$.

```python
# Questão f:
chi2_calc = 17.60
chi2_crit = 19.6751

if chi2_calc > chi2_crit:
    print("Decisão: Rejeitar H0")
else:
    print("Decisão: Não rejeitar H0")
```
