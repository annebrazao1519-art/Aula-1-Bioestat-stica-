```python
#Pergunta 4
##Variabilidade aparente dos tratamentoss
variabilidade = df.groupby('Tratamento')['Brotos'].agg(
    Desvio_Padrao='std',
    Amplitude=lambda x: x.max() - x.min(),
    IQR=lambda x: x.quantile(0.75) - x.quantile(0.25)
)

print(variabilidade)
```

#Interpretação dos dados obtidos para responder a pergunta 4
##a parti do dos dados obtidos através do cálculo do desvio padrão, amplitude e da IQR é possível identificar o tratamento que possui maior variabilidade. Isso porque a variabilidade é definida pelo grau de dispersão dos dados em relação à concentração de dados. A amplitude total vai medir a distância bruta entre o valor máximo e mínimo observado ao decorrer do tratamento. O IQR indica a dispersão central dos dados (50%), ao subtrair o q3-q1 e o desvio indica em média o quanto as observações se distanciam da média do grupo de dados.

##Assim, o tratamento 2, que apresenta os maiores valores ao mesmo tempo nessas medidas, é o que possui maior variabilidade de respostas ao tratamento do BAP.

##Na análise dessa variável em questão, o tratamento tem queda no desenvolvimento da folhagem. Sendo a sua mediana = 11 folhas, mais baixo que o do tratamento 1, por exemplo.

```python
def amplitude(x):
    return x.max() - x.min()

estatisticas = df.groupby('Tratamento').agg(
    Brotos_Amplitude=('Brotos', amplitude),
    Brotos_Variancia=('Brotos', 'var'),
    Brotos_Desvio_Padrao=('Brotos', 'std'),
    Folhas_Amplitude=('Folhas', amplitude),
    Folhas_Variancia=('Folhas', 'var'),
    Folhas_Desvio_Padrao=('Folhas', 'std')
)

print(estatisticas.round(2))
```

```python
#Pergunta 9

def cv(x):
    return (x.std() / x.mean()) * 100

df_cv = df.groupby('Tratamento')[['Brotos', 'Folhas']].apply(cv)

print(df_cv)
```
