#***Distribuição normal***

##É um modelo de probabilidade contínuo onde a média, a mediana e a moda coincidem no centro, sendo de formato simétrico. Usamos então a Função Densidade de Probabilidade (PDF) que é a fórmula matemática que descreve onde os valores de uma variável contínua (como a largura de uma sépala em centímetros) têm maior tendência a se aglomerar.
##$$f(x; \mu, \sigma) = \frac{1}{\sigma \sqrt{2\pi}} \exp\left( -\frac{1}{2} \left( \frac{x - \mu}{\sigma} \right)^2 \right), \quad \text{para } x \in \mathbb{R}$$
##Definição dos Símbolos:
##$x$: Variável aleatória contínua representando a medida biológica (largura da sépala em cm).
##$\mu$: Média populacional
##($\mu \in \mathbb{R}$), parâmetro de posição ou localização.
##$\sigma$: Desvio padrão populacional
##($\sigma > 0$), parâmetro de escala ou dispersão ($\sigma^2$ é a variância).
##$\pi$: Constante matemática de Arquimedes ($\pi \approx 3{,}14159$).
##$e$: Base do logaritmo natural ou número de Euler ($e \approx 2{,}71828$).


####Referência consultada
####BUSSAB, Wilton de O.; MORETTIN, Pedro A. Estatística Básica. 9. ed. São Paulo: Saraiva, 2017.


#O que essa distribuição representa?
##A distribuição normal foi utilizada como um modelo de parâmetro para a medição individual de um organimos. Essa distribuição representa uma variação morfológica esperada entre os indivíduos da população, onde sofre a influência de múltiplos fatores genéticos ou amienais. Ou seja, ela não representa uma média de várias amostras ou razão entre as variâncias.


#Identificação dos Parâmetros, Valores Admissíveis e Efeitos na Curva
##A Distribuição Normal é completamente determinada por dois parâmetros: a média ($\mu$) e o desvio-padrão ($\sigma$).

##A média ($\mu$) tem como valores admissíveis qualquer número real. Quando alterado esse valor positivamente, acontece a movimentação do eixo de simetria e o ápice da curva para direita e quando $\mu$ diminui, move-se para esquerda

##Já o desvio padrão ($\sigma$) precisa ter um valor maior do que 0 para ser admissível. Esse parâmetro conrola o grau de dispersão, valores maiores tendem a achatar o topo da curva e alargar as caudas, enquanto valores menores tornam as curvas mais afuniladas e concentradas ao redor da média



```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Configuração da figura
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
x = np.linspace(1, 6, 500)

# 1. Variando mu (sigma fixo em 0.4)
sigma_fixo = 0.4
for mu in [2.5, 3.4, 4.3]:
    pdf = norm.pdf(x, loc=mu, scale=sigma_fixo)
    axes[0].plot(x, pdf, label=f'$\mu = {mu}$, $\sigma = {sigma_fixo}$')
axes[0].set_title('Variação do Parâmetro de Posição ($\mu$)')
axes[0].set_xlabel('Largura da Sépala (cm)')
axes[0].set_ylabel('Densidade f(x)')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# 2. Variando sigma (mu fixo em 3.4)
mu_fixo = 3.4
for sigma in [0.2, 0.4, 0.7]:
    pdf = norm.pdf(x, loc=mu_fixo, scale=sigma)
    axes[1].plot(x, pdf, label=f'$\mu = {mu_fixo}$, $\sigma = {sigma}$')
axes[1].set_title('Variação do Parâmetro de Escala ($\sigma$)')
axes[1].set_xlabel('Largura da Sépala (cm)')
axes[1].set_ylabel('Densidade f(x)')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

#Discussão de aplicabilidade
##Nesse exemplo, é possível usar o Modelo Normal para analisar as flores de Iris, pois os dados seguem os requisitos básicos que são:
##As flores possuem uma medida contínua, onde é medida em centímetro e pode assumir qualquer valor real; E os valores são simétricos, onde a maioria das flores possuem tamanhos muito próximos da média, e outlier são raros de se encontrar.

##Porém na teoria matemática, diz que uma curva perfeita vai do infinito negativo ($-\infty$) ao infinito positivo ($+\infty$), porém não existem flores com valores negativos de comprimento.
##Mas podemos ignorar esse fato e utilizar a distribuição normal nesse caso porque a média das flores é de 3.43 cm, e o desvio padrão é de 0,38 cm, sendo assim, para que uma flor tenha flores negativos ou zerado, ela precisa de um desvio padrão maior do que 9 de distância da média.




#Contextualização Prática
##Para fim didático, podemos fazer algumas perguntas para serem respondidas com base nos valores dos gráficos.
##1.Menor ou igual a 3,0 cm? — $P(X \le 3{,}0)$
##2.Entre 3,0 cm e 3,8 cm? — $P(3{,}0 \le X \le 3{,}8)$
##3.Maior que 3,8 cm? — $P(X > 3{,}8)$
##4.Qual é a largura máxima de sépala que compreende os 90% menores indivíduos? (90º percentil)

```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 1. Carregar dados reais
df = sns.load_dataset('iris')
setosa_sw = df[df['species'] == 'setosa']['sepal_width']

# 2. Estimar parâmetros amostrais da Iris setosa
mu_est = setosa_sw.mean()
sigma_est = setosa_sw.std(ddof=1)

print(f"Média amostral (mu): {mu_est:.4f} cm")
print(f"Desvio-padrão amostral (sigma): {sigma_est:.4f} cm")

# 3. Cálculos de Probabilidade com SciPy
limit_low = 3.0
limit_high = 3.8

# P(X <= 3.0) usando CDF
p_cdf_low = norm.cdf(limit_low, loc=mu_est, scale=sigma_est)

# P(3.0 <= X <= 3.8) usando diferença de CDFs
p_interval = norm.cdf(limit_high, loc=mu_est, scale=sigma_est) - norm.cdf(limit_low, loc=mu_est, scale=sigma_est)

# P(X > 3.8) usando Survival Function (SF = 1 - CDF)
p_sf_high = norm.sf(limit_high, loc=mu_est, scale=sigma_est)

# 90º Percentil usando PPF (Percent Point Function)
p90 = norm.ppf(0.90, loc=mu_est, scale=sigma_est)

print(f"\n--- Resultados de Probabilidade ---")
print(f"P(X <= {limit_low} cm) = {p_cdf_low:.4f} ({p_cdf_low*100:.2f}%)")
print(f"P({limit_low} <= X <= {limit_high} cm) = {p_interval:.4f} ({p_interval*100:.2f}%)")
print(f"P(X > {limit_high} cm) = {p_sf_high:.4f} ({p_sf_high*100:.2f}%)")
print(f"90% das flores de Setosa possuem largura de sépala <= {p90:.2f} cm")

# 4. Geração dos Gráficos com Áreas Sombreadas
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
x_axis = np.linspace(mu_est - 4*sigma_est, mu_est + 4*sigma_est, 500)
pdf_vals = norm.pdf(x_axis, loc=mu_est, scale=sigma_est)

# Gráfico 1: P(X <= 3.0)
axes[0].plot(x_axis, pdf_vals, 'k-', lw=2)
x_fill1 = x_axis[x_axis <= limit_low]
axes[0].fill_between(x_fill1, norm.pdf(x_fill1, mu_est, sigma_est), color='skyblue', alpha=0.6, label=f'P(X ≤ 3.0) = {p_cdf_low:.4f}')
axes[0].axvline(limit_low, color='blue', linestyle='--')
axes[0].set_title('Probabilidade Acumulada à Esquerda (CDF)')
axes[0].set_xlabel('Largura da Sépala (cm)')
axes[0].set_ylabel('Densidade f(x)')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Gráfico 2: P(3.0 <= X <= 3.8)
axes[1].plot(x_axis, pdf_vals, 'k-', lw=2)
x_fill2 = x_axis[(x_axis >= limit_low) & (x_axis <= limit_high)]
axes[1].fill_between(x_fill2, norm.pdf(x_fill2, mu_est, sigma_est), color='limegreen', alpha=0.6, label=f'P(3.0 ≤ X ≤ 3.8) = {p_interval:.4f}')
axes[1].axvline(limit_low, color='green', linestyle='--')
axes[1].axvline(limit_high, color='green', linestyle='--')
axes[1].set_title('Probabilidade em Intervalo')
axes[1].set_xlabel('Largura da Sépala (cm)')
axes[1].set_ylabel('Densidade f(x)')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

# Gráfico 3: P(X > 3.8) e Percentil 90
axes[2].plot(x_axis, pdf_vals, 'k-', lw=2)
x_fill3 = x_axis[x_axis >= limit_high]
axes[2].fill_between(x_fill3, norm.pdf(x_fill3, mu_est, sigma_est), color='salmon', alpha=0.6, label=f'P(X > 3.8) = {p_sf_high:.4f}')
axes[2].axvline(p90, color='darkred', linestyle=':', lw=2, label=f'P90 = {p90:.2f} cm')
axes[2].set_title('Cauda Superior (SF) e 90º Percentil')
axes[2].set_xlabel('Largura da Sépala (cm)')
axes[2].set_ylabel('Densidade f(x)')
axes[2].legend()
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

#***Distribuição t de Student***

##A Distribuição $t$ de Student é uma distribuição de probabilidade contínua e simétrica, semelhante a distribuição normal, mas apresenta as caudas maiores, com maior probabilidade de valores extremos. Ela emerge do estudo do comportamento de médias amostrais quando o desvio-padrão populacional ($\sigma$) é desconhecido e precisa ser estimado pelo desvio-padrão amostral ($s$).
## A fórmula é dada por:
##$$f(t; \nu) = \frac{\Gamma\left(\frac{\nu+1}{2}\right)}{\sqrt{\nu\pi} \, \Gamma\left(\frac{\nu}{2}\right)} \left(1 + \frac{t^2}{\nu}\right)^{-\frac{\nu+1}{2}}, \quad \text{para } t \in \mathbb{R}$$
##Onde:
##- $t$: Estatística amostral padronizada ($t \in \mathbb{R}$), definida como:$$t = \frac{\bar{X} - \mu_0}{\frac{s}{\sqrt{n}}}$$
##onde $\bar{X}$ é a média da amostra, $\mu_0$ é a média populacional de referência (hipotética), $s$ é o desvio-padrão amostral e $n$ é o tamanho da amostra.
##- $\nu$ (ou $df$): Graus de liberdade ($\nu = n - 1$) para uma amostra simples, com $\nu > 0$).
##- $\Gamma(\cdot)$: Função Gama, função de interpolação do fatorial para números reais ($\Gamma(k) = (k-1)!$ para inteiros positivos).
##- $\pi$: Constante matemática ($\approx 3{,}14159$).

#Referência consultada
##MORETTIN, Pedro A.; BUSSAB, Wilton de O. Estatística Básica. 9. ed. São Paulo: Saraiva, 2017.

#O que essa distribuição representa?

##Enquanto a distribuição normal diz respeita a medição de 1 único indivíduo, a distribuição t trabalha com o comportamento da média de um lote. Como não sabemos qual é a média do tamanho de todas as flores de Iris do planeta, pegamos o desvio padrão da nossa amostra para resolver esse problema, mas isso ocasiona em uma estimativa impefeita, fazendo com que as amostras ocilem.
##A distribuição t é uma ferramenta para compensar exatamente esse erro de estimativa.

#Identificação dos Parâmetros, Valores Admissíveis e Efeitos na Curva
##O único parâmetro que define o formato da distribuição $t$ central é o grau de liberdade ($\nu$ ou $df$).
##Para valores pequenos de $\nu$ (amostras pequenas), o pico da curva é mais baixo e as caudas são mais elevadas/pesadas, refletindo a maior incerteza na estimativa de $s$.
##Conforme $\nu \to \infty$ (amostras grandes), a variabilidade na estimativa de $s$ desaparece e a distribuição $t$ converge assintoticamente para a Distribuição Normal Padrão $Z \sim \mathcal{N}(0, 1)$.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t, norm

# Configuração da figura com dois painéis
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Eixo t principal e eixo t focado na cauda
t_axis = np.linspace(-4, 4, 500)
t_tail = np.linspace(1.5, 4, 300)

# Graus de liberdade com alto contraste:
# 1 (extremamente instável), 3 (amostra pequena), 10 (intermediária), 49 (nossa amostra da Iris)
dfs = [1, 3, 10, 49]
colors = ['#d62728', '#ff7f0e', '#2ca02c', '#1f77b4']

# -------------------------------------------------------------
# PAINEL 1: Visão Geral (Centro e Altura do Pico)
# -------------------------------------------------------------
for df_val, color in zip(dfs, colors):
    ax1.plot(t_axis, t.pdf(t_axis, df=df_val), label=rf'$\nu = {df_val}$', color=color, lw=2.2)

# Normal Padrão (Z) como referência limite (\nu -> infinito)
ax1.plot(t_axis, norm.pdf(t_axis), 'k--', lw=2.5, label=r'Normal Padrão $Z \sim \mathcal{N}(0,1)$')

ax1.set_title(r'Visão Geral: Centro e Altura do Pico', fontsize=12, fontweight='bold')
ax1.set_xlabel('Estatística $t$ (adimensional)', fontsize=11)
ax1.set_ylabel('Densidade de Probabilidade $f(t)$', fontsize=11)
ax1.grid(True, alpha=0.3)
ax1.legend(fontsize=10)

# -------------------------------------------------------------
# PAINEL 2: Zoom na Cauda Direita (Destaque para as Caudas Pesadas)
# -------------------------------------------------------------
for df_val, color in zip(dfs, colors):
    ax2.plot(t_tail, t.pdf(t_tail, df=df_val), label=rf'$\nu = {df_val}$', color=color, lw=2.2)

ax2.plot(t_tail, norm.pdf(t_tail), 'k--', lw=2.5, label=r'Normal Padrão $Z \sim \mathcal{N}(0,1)$')

ax2.set_title(r'Destaque nas Caudas: Probabilidade de Outliers ($t > 1.5$)', fontsize=12, fontweight='bold')
ax2.set_xlabel('Estatística $t$ (adimensional)', fontsize=11)
ax2.set_ylabel('Densidade de Probabilidade $f(t)$', fontsize=11)
ax2.grid(True, alpha=0.3)
ax2.legend(fontsize=10)

fig.suptitle(r'Efeito dos Graus de Liberdade ($\nu$) na Distribuição $t$ de Student', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

#Discussão de aplicabilidade
##A variável biológica de origem em cada grupo deve ser aproximadamente Normal, enquando as observações amostrais devem ser independentes umas das outras.
##Por exemplo, temos $n = 50$ flores, portanto os graus de liberdade são $\nu = 50 - 1 = 49$.
##Com $\nu = 49$, a curva $t$ já é extremamente próxima da Normal Padrão, fornecendo estimativas altamente robustas para a média populacional da espécie.

#Contextualização Prática
##Como a distribuição t de Studant é uma média de uma amostra de população, por exemplo, o pesquisador entrou no mato e coletou 50 flores de Iris aleatóriamente, calculando assim a média do grupo. Como é um valor distante de 1 amostra, a média tende a flutar menos, já que os valores maiores se anulam com os valores menores, se aproximando da média
##Então, pergunta-se
##Imagine que um estudo prévio na literatura afirme que a largura média populacional da sépala de Iris setosa é $\mu_0 = 3{,}35$ cm. Se coletarmos uma amostra aleatória de $n = 50$ flores com desvio-padrão $s = 0{,}3791$ cm, qual é a distribuição da estatística $t$ padronizada sob essa hipótese e qual a probabilidade de obtermos:
##Uma estatística $t \le -1{,}50$? (Média amostral $\bar{X} \le 3{,}270$ cm)
##Uma estatística $t$ entre $-1{,}50$ e $+1{,}50$? (Média amostral entre $3{,}270$ cm e $3{,}430$ cm)
##Uma estatística $t > +1{,}50$? (Média amostral $\bar{X} > 3{,}430$ cm)
##Qual é o $90^\circ$ percentil da estatística $t$ e a qual média amostral em centímetros ele equivale?

```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t

# 1. Obter dados da Iris setosa
df = sns.load_dataset('iris')
setosa_sw = df[df['species'] == 'setosa']['sepal_width']

n = len(setosa_sw)          # n = 50
df_val = n - 1             # degrees of freedom = 49
s_est = setosa_sw.std(ddof=1)  # s = 0.3791 cm
se_est = s_est / np.sqrt(n)     # SE = 0.05361 cm

mu_ref = 3.35  # Média de referência biológica (em cm)

print(f"Tamanho da amostra (n): {n}")
print(f"Graus de Liberdade (df): {df_val}")
print(f"Desvio-padrão amostral (s): {s_est:.4f} cm")
print(f"Erro-padrão da média (SE): {se_est:.4f} cm")

# 2. Definir limites na escala t e converter para cm
t_low = -1.50
t_high = 1.50

# Conversão da estatística t para a escala original em centímetros (X_bar = mu_0 + t * SE)
x_bar_low = mu_ref + t_low * se_est
x_bar_high = mu_ref + t_high * se_est

print(f"\n--- Conversão t -> cm (com mu_0 = {mu_ref} cm) ---")
print(f"t = {t_low:.2f}  <=>  Média Amostral = {x_bar_low:.4f} cm")
print(f"t = {t_high:.2f}  <=>  Média Amostral = {x_bar_high:.4f} cm")

# 3. Cálculos de Probabilidade com SciPy (scipy.stats.t)
p_cdf_low = t.cdf(t_low, df=df_val)
p_interval = t.cdf(t_high, df=df_val) - t.cdf(t_low, df=df_val)
p_sf_high = t.sf(t_high, df=df_val)

# 90º Percentil de t
t_p90 = t.ppf(0.90, df=df_val)
x_bar_p90 = mu_ref + t_p90 * se_est

print(f"\n--- Resultados da Distribuição t (df = {df_val}) ---")
print(f"P(T <= {t_low}) = {p_cdf_low:.4f} ({p_cdf_low*100:.2f}%)")
print(f"P({t_low} <= T <= {t_high}) = {p_interval:.4f} ({p_interval*100:.2f}%)")
print(f"P(T > {t_high}) = {p_sf_high:.4f} ({p_sf_high*100:.2f}%)")
print(f"90º Percentil t = {t_p90:.4f} <=> Média Amostral max <= {x_bar_p90:.4f} cm")

# 4. Geração dos Gráficos com Áreas Sombreadas
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
t_axis = np.linspace(-4, 4, 500)
pdf_vals = t.pdf(t_axis, df=df_val)

# Gráfico 1: P(T <= -1.5)
axes[0].plot(t_axis, pdf_vals, 'k-', lw=2)
t_fill1 = t_axis[t_axis <= t_low]
axes[0].fill_between(t_fill1, t.pdf(t_fill1, df=df_val), color='skyblue', alpha=0.6, label=f'P(T ≤ -1.50) = {p_cdf_low:.4f}')
axes[0].axvline(t_low, color='blue', linestyle='--')
axes[0].set_title('Cauda Inferior (CDF)')
axes[0].set_xlabel('Estatística t (adimensional)')
axes[0].set_ylabel('Densidade f(t)')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Gráfico 2: P(-1.5 <= T <= 1.5)
axes[1].plot(t_axis, pdf_vals, 'k-', lw=2)
t_fill2 = t_axis[(t_axis >= t_low) & (t_axis <= t_high)]
axes[1].fill_between(t_fill2, t.pdf(t_fill2, df=df_val), color='limegreen', alpha=0.6, label=f'P(-1.50 ≤ T ≤ 1.50) = {p_interval:.4f}')
axes[1].axvline(t_low, color='green', linestyle='--')
axes[1].axvline(t_high, color='green', linestyle='--')
axes[1].set_title('Intervalo Central de t')
axes[1].set_xlabel('Estatística t (adimensional)')
axes[1].set_ylabel('Densidade f(t)')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

# Gráfico 3: P(T > 1.5) e Percentil 90
axes[2].plot(t_axis, pdf_vals, 'k-', lw=2)
t_fill3 = t_axis[t_axis >= t_high]
axes[2].fill_between(t_fill3, t.pdf(t_fill3, df=df_val), color='salmon', alpha=0.6, label=f'P(T > 1.50) = {p_sf_high:.4f}')
axes[2].axvline(t_p90, color='darkred', linestyle=':', lw=2, label=f'P90 (t={t_p90:.2f})')
axes[2].set_title('Cauda Superior (SF) e 90º Percentil')
axes[2].set_xlabel('Estatística t (adimensional)')
axes[2].set_ylabel('Densidade f(t)')
axes[2].legend()
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

#***Distribuição Qui-Quadrado***
##A Distribuição Qui-quadrado é uma distribuição de probabilidade contínua, assimétrica à direita e definida apenas para valores não negativos ($x \ge 0$).
##Ela resulta da soma dos quadrados de $k$ variáveis aleatórias independentes e padronizadas com Distribuição Normal Padrão $Z \sim \mathcal{N}(0, 1)$.
##A Função Densidade de Probabilidade é dada por:
##$$f(x; k) = \begin{cases} \frac{1}{2^{k/2} \, \Gamma\left(\frac{k}{2}\right)} \, x^{\left(\frac{k}{2} - 1\right)} e^{-\frac{x}{2}}, & \text{para } x > 0 \\ 0, & \text{para } x \le 0 \end{cases}$$
##Onde:
##$x$ (ou $Q$): Variável aleatória contínua não negativa ($x \ge 0$), representando a estatística de variância padronizada:
##$$Q = \frac{(n - 1)s^2}{\sigma_0^2}$$
##onde $s^2$ é a variância amostral calculada, $n$ é o tamanho da amostra e $\sigma_0^2$ é a variância populacional de referência (hipotética).
##$k$ (ou $\nu$): Graus de liberdade ($k = n - 1$ para a variância amostral de um grupo, $k \in \mathbb{N}^*$).
##$\Gamma(\cdot)$: Função Gama.
##$e$: Base do logaritmo natural ($\approx 2{,}71828$).

##Referência consultada
##BUSSAB, Wilton de O.; MORETTIN, Pedro A. Estatística Básica. 9. ed. São Paulo: Saraiva, 2017.

#O que essa distribuição representa?
##A distribuição $\chi^2$ modela uma estatística calculada a partir de uma amostra, especificamente a soma de desvios quadráticos ou a variância amostral escalonada ($s^2$).
##Medida biológica individual: $X_i$ (largura da sépala em cm).
##Variância amostral: $s^2 = \frac{1}{n-1}\sum_{i=1}^{n} (X_i - \bar{X})^2$ (em $\text{cm}^2$).
##Estatística $\chi^2$: $Q = \frac{(n-1)s^2}{\sigma_0^2}$ (adimensional), que mede o quão distante a variância amostral observada está de uma variância populacional esperada $\sigma_0^2$.

#Identificação dos Parâmetros, Valores Admissíveis e Efeitos na Curva
##O único parâmetro que define a distribuição Qui-quadrado é o número de graus de liberdade ($k$).
##Para $k$ pequeno ($k = 2, 3, 5$), a curva é fortemente assimétrica à direita (com assimetria positiva marcante) e concentrada próxima à origem.
##O valor esperado (média) da distribuição é igual a $k$ ($\mathbb{E}[X] = k$) e a sua variância é $2k$ ($\text{Var}(X) = 2k$).
##Conforme $k$ aumenta, a curva se desloca para a direita, alarga-se e torna-se gradualmente mais simétrica, aproximando-se da forma de sino (pelo Teorema Central do Limite).






```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import chi2

# Eixo x para os valores da estatística Qui-quadrado
x_axis = np.linspace(0, 80, 500)

plt.figure(figsize=(10, 5))

# Variando os graus de liberdade k
ks = [2, 5, 15, 49]
colors = ['purple', 'orange', 'green', 'blue']

for k_val, color in zip(ks, colors):
    pdf_chi2 = chi2.pdf(x_axis, df=k_val)
    plt.plot(x_axis, pdf_chi2, label=f'$\chi^2$ ($k = {k_val}$)', color=color, lw=1.8)

plt.title('Efeito dos Graus de Liberdade ($k$) na Distribuição Qui-quadrado')
plt.xlabel('Estatística $\chi^2$ (adimensional)')
plt.ylabel('Densidade $f(x)$')
plt.ylim(0, 0.25)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

#Contextualização Prática
##Suponha que um modelo teórico de genética de populações preveja que a variância da largura da sépala em Iris setosa seja $\sigma_0^2 = 0{,}1200\text{ cm}^2$ ($\sigma_0 \approx 0{,}3464\text{ cm}$).
##Ao coletar uma amostra de $n = 50$ flores ($k = 49$), qual é a probabilidade de a estatística Qui-quadrado amostral $Q = \frac{49 \cdot s^2}{0{,}12}$ resultar em:
##Uma estatística $Q \le 35{,}0$? (Variância amostral $s^2 \le 0{,}0857\text{ cm}^2$)
##Uma estatística $Q$ no intervalo entre $35{,}0$ e $65{,}0$? (Variância $0{,}0857\text{ cm}^2 \le s^2 \le 0{,}1592\text{ cm}^2$)
##Uma estatística $Q > 65{,}0$? (Variância amostral $s^2 > 0{,}1592\text{ cm}^2$)
##Qual é o $90^\circ$ percentil de $Q$ e a qual variância amostral em $\text{cm}^2$ ele corresponde?

```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import chi2

# 1. Obter dados da Iris setosa
df = sns.load_dataset('iris')
setosa_sw = df[df['species'] == 'setosa']['sepal_width']

n = len(setosa_sw)             # n = 50
k_val = n - 1                 # k = 49
s2_obs = setosa_sw.var(ddof=1) # s^2 observado = 0.1437 cm^2

sigma0_sq = 0.1200            # Variância hipotética de referência (cm^2)
q_calculated = (k_val * s2_obs) / sigma0_sq

print(f"Tamanho da amostra (n): {n}")
print(f"Graus de Liberdade (k): {k_val}")
print(f"Variância amostral observada (s²): {s2_obs:.4f} cm²")
print(f"Estatística Q (Qui-quadrado calculada): {q_calculated:.4f}")

# 2. Definir limites na escala Q e converter para a escala biológica (s² em cm²)
q_low = 35.0
q_high = 65.0

s2_low = (q_low * sigma0_sq) / k_val
s2_high = (q_high * sigma0_sq) / k_val

print(f"\n--- Conversão Q -> s² (cm²) com sigma0² = {sigma0_sq} cm² ---")
print(f"Q = {q_low:.1f}  <=>  Variância Amostral s² = {s2_low:.4f} cm²")
print(f"Q = {q_high:.1f}  <=>  Variância Amostral s² = {s2_high:.4f} cm²")

# 3. Cálculos de Probabilidade com SciPy (scipy.stats.chi2)
p_cdf_low = chi2.cdf(q_low, df=k_val)
p_interval = chi2.cdf(q_high, df=k_val) - chi2.cdf(q_low, df=k_val)
p_sf_high = chi2.sf(q_high, df=k_val)

# 90º Percentil de Q
q_p90 = chi2.ppf(0.90, df=k_val)
s2_p90 = (q_p90 * sigma0_sq) / k_val

print(f"\n--- Resultados da Distribuição Qui-quadrado (k = {k_val}) ---")
print(f"P(Q <= {q_low}) = {p_cdf_low:.4f} ({p_cdf_low*100:.2f}%)")
print(f"P({q_low} <= Q <= {q_high}) = {p_interval:.4f} ({p_interval*100:.2f}%)")
print(f"P(Q > {q_high}) = {p_sf_high:.4f} ({p_sf_high*100:.2f}%)")
print(f"90º Percentil Q = {q_p90:.4f} <=> Variância Amostral max s² <= {s2_p90:.4f} cm²")

# 4. Geração dos Gráficos com Áreas Sombreadas
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
q_axis = np.linspace(15, 85, 500)
pdf_vals = chi2.pdf(q_axis, df=k_val)

# Gráfico 1: P(Q <= 35.0)
axes[0].plot(q_axis, pdf_vals, 'k-', lw=2)
q_fill1 = q_axis[q_axis <= q_low]
axes[0].fill_between(q_fill1, chi2.pdf(q_fill1, df=k_val), color='skyblue', alpha=0.6, label=f'P(Q ≤ 35.0) = {p_cdf_low:.4f}')
axes[0].axvline(q_low, color='blue', linestyle='--')
axes[0].set_title('Cauda Inferior (CDF)')
axes[0].set_xlabel('Estatística Q ($\chi^2$)')
axes[0].set_ylabel('Densidade f(Q)')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Gráfico 2: P(35.0 <= Q <= 65.0)
axes[1].plot(q_axis, pdf_vals, 'k-', lw=2)
q_fill2 = q_axis[(q_axis >= q_low) & (q_axis <= q_high)]
axes[1].fill_between(q_fill2, chi2.pdf(q_fill2, df=k_val), color='limegreen', alpha=0.6, label=f'P(35.0 ≤ Q ≤ 65.0) = {p_interval:.4f}')
axes[1].axvline(q_low, color='green', linestyle='--')
axes[1].axvline(q_high, color='green', linestyle='--')
axes[1].set_title('Intervalo Central de Q')
axes[1].set_xlabel('Estatística Q ($\chi^2$)')
axes[1].set_ylabel('Densidade f(Q)')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

# Gráfico 3: P(Q > 65.0) e Percentil 90
axes[2].plot(q_axis, pdf_vals, 'k-', lw=2)
q_fill3 = q_axis[q_axis >= q_high]
axes[2].fill_between(q_fill3, chi2.pdf(q_fill3, df=k_val), color='salmon', alpha=0.6, label=f'P(Q > 65.0) = {p_sf_high:.4f}')
axes[2].axvline(q_p90, color='darkred', linestyle=':', lw=2, label=f'P90 (Q={q_p90:.2f})')
axes[2].set_title('Cauda Superior (SF) e 90º Percentil')
axes[2].set_xlabel('Estatística Q ($\chi^2$)')
axes[2].set_ylabel('Densidade f(Q)')
axes[2].legend()
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

#***Distribuição f de Fisher***
##A Distribuição $F$ é uma distribuição de probabilidade contínua, não negativa ($F > 0$) e assimétrica à direita. Ela surge como a distribuição da razão entre duas variáveis aleatórias independentes com distribuição Qui-quadrado, divididas pelos seus respectivos graus de liberdade.
##A Função Densidade de Probabilidade é dada por:
##$$f(F; d_1, d_2) = \frac{\Gamma\left(\frac{d_1 + d_2}{2}\right) \left(\frac{d_1}{d_2}\right)^{\frac{d_1}{2}} F^{\frac{d_1}{2} - 1}}{\Gamma\left(\frac{d_1}{2}\right) \Gamma\left(\frac{d_2}{2}\right) \left(1 + \frac{d_1}{d_2} F\right)^{\frac{d_1 + d_2}{2}}}, \quad \text{para } F > 0$$
##Onde:
##$F$: Estatística amostral ($F > 0$), definida como a razão entre duas variâncias amostrais de populações normais sob a hipótese de igualdade de variâncias populacionais ($\sigma_1^2 = \sigma_2^2$):
##$$F = \frac{s_1^2}{s_2^2}$$
##$d_1$ (ou $df_1$, $\nu_1$): Graus de liberdade do numerador ($d_1 = n_1 - 1$, com $d_1 \in \mathbb{N}^*$).
##$d_2$ (ou $df_2$, $\nu_2$): Graus de liberdade do denominador ($d_2 = n_2 - 1$, com $d_2 \in \mathbb{N}^*$).
##$\Gamma(\cdot)$: Função Gama.

##Referência consultada
##BUSSAB, Wilton de O.; MORETTIN, Pedro A. Estatística Básica. 9. ed. São Paulo: Saraiva, 2017.

#O que essa distribuição representa?
## A distribuição $F$ modela exclusivamente a razão entre duas estimativas independentes de variância amostral ($s_1^2 / s_2^2$), logo, nenhuma variável morfológica individual em centímetros possui distribuição $F$.
##Dando por exemplo:
##Amostra 1 (Iris setosa): $n_1 = 50$, variância amostral $s_1^2$ (em $\text{cm}^2$).
##Amostra 2 (Iris versicolor): $n_2 = 50$, variância amostral $s_2^2$ (em $\text{cm}^2$).Estatística $F$: $F = \frac{s_1^2}{s_2^2}$ (adimensional).
##Se as duas espécies possuírem exatamente a mesma variância populacional ($\sigma_1^2 = \sigma_2^2$), o valor esperado de $F$ estará próximo de $1$.
## Desvios significativos acima de $1$ indicam maior variabilidade no primeiro grupo.

#Identificação dos Parâmetros, Valores Admissíveis e Efeitos na Curva
##A forma da distribuição $F$ é controlada separadamente pelos graus de liberdade do numerador ($df_1$) e do denominador ($df_2$).
##Investigação de $df_1$ (Numerador): Para $df_1$ pequeno (ex.: $df_1 = 2$), a densidade é extremamente assimétrica e decresce a partir do zero. Conforme $df_1$ aumenta, a curva desenvolve um pico definido que se desloca para a direita.
##Investigação de $df_2$ (Denominador): Controla o peso da cauda direita e a variância do modelo ($\text{Var}(F)$ é inversamente proporcional a $df_2$). Para $df_2$ pequenos, a cauda direita é pesada, permitindo valores de $F$ muito elevados. Conforme $df_2 \to \infty$, a dispersão diminui e a curva se afunila ao redor de $1$.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import f

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
f_axis = np.linspace(0.01, 4, 500)

# 1. Variando df1 (df2 fixo em 49)
df2_fixed = 49
for df1_val in [2, 5, 15, 49]:
    pdf_f = f.pdf(f_axis, dfn=df1_val, dfd=df2_fixed)
    axes[0].plot(f_axis, pdf_f, label=f'$df_1 = {df1_val}, df_2 = {df2_fixed}$')
axes[0].set_title('Efeito dos Graus de Liberdade do Numerador ($df_1$)')
axes[0].set_xlabel('Estatística F (adimensional)')
axes[0].set_ylabel('Densidade f(F)')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# 2. Variando df2 (df1 fixo em 49)
df1_fixed = 49
for df2_val in [2, 5, 15, 49]:
    pdf_f = f.pdf(f_axis, dfn=df1_fixed, dfd=df2_val)
    axes[1].plot(f_axis, pdf_f, label=f'$df_1 = {df1_fixed}, df_2 = {df2_val}$')
axes[1].set_title('Efeito dos Graus de Liberdade do Denominador ($df_2$)')
axes[1].set_xlabel('Estatística F (adimensional)')
axes[1].set_ylabel('Densidade f(F)')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

#Discussão de aplicabilidade
##Para ter aplicabilidade, as condições necessárias são:
##A variável morfológica deve seguir uma distribuição aproximadamente Normal em ambas as populações de origem.
##As duas amostras devem ser biologicamente independentes entre si
## O teste $F$ para razões de variância é notoriamente sensível a desvios da Normalidade.  Mas a base Iris, a largura da sépala em ambas as espécies apresenta boa simetria, torna o modelo viável para uso.

#Contextualização Prática
##A variabilidade na largura da sépala de Iris setosa ($s_1^2$) é significativamente maior do que a de Iris versicolor ($s_2^2$)? Sob a hipótese nula de que ambas as espécies possuem a mesma variância populacional ($\sigma_1^2 = \sigma_2^2$), qual é a probabilidade de obtermos uma razão de variâncias $F = \frac{s_{\text{setosa}}^2}{s_{\text{versicolor}}^2}$ (com $df_1 = 49$ e $df_2 = 49$):
##Menor ou igual a $0{,}70$? — $P(F \le 0{,}70)$
##No intervalo entre $0{,}70$ e $1{,}50$? — $P(0{,}70 \le F \le 1{,}50)$
##Maior que $1{,}50$? — $P(F > 1{,}50)$
##Qual é o $90^\circ$ percentil crítico da distribuição $F(49, 49)$ e como ele se compara com a razão observada nos dados reais?

```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import f

# 1. Carregar e separar os dados das duas espécies
df = sns.load_dataset('iris')
setosa_sw = df[df['species'] == 'setosa']['sepal_width']
versicolor_sw = df[df['species'] == 'versicolor']['sepal_width']

n1, n2 = len(setosa_sw), len(versicolor_sw)
df1, df2 = n1 - 1, n2 - 1  # df1 = 49, df2 = 49

var_setosa = setosa_sw.var(ddof=1)       # s1^2 = 0.1437 cm^2
var_versicolor = versicolor_sw.var(ddof=1) # s2^2 = 0.0985 cm^2

f_obs = var_setosa / var_versicolor       # F calculado = 1.4592

print(f"Iris setosa: n1 = {n1}, s1² = {var_setosa:.4f} cm²")
print(f"Iris versicolor: n2 = {n2}, s2² = {var_versicolor:.4f} cm²")
print(f"Razão de Variâncias Amostrais (F observado): {f_obs:.4f}")

# 2. Limites para cálculo de probabilidades
f_low = 0.70
f_high = 1.50

# 3. Cálculos de Probabilidade com SciPy (scipy.stats.f)
p_cdf_low = f.cdf(f_low, dfn=df1, dfd=df2)
p_interval = f.cdf(f_high, dfn=df1, dfd=df2) - f.cdf(f_low, dfn=df1, dfd=df2)
p_sf_high = f.sf(f_high, dfn=df1, dfd=df2)

# Probabilidade de obter um F igual ou maior que o observado nos dados reais
p_value_f_obs = f.sf(f_obs, dfn=df1, dfd=df2)

# 90º e 95º Percentis de F
f_p90 = f.ppf(0.90, dfn=df1, dfd=df2)
f_p95 = f.ppf(0.95, dfn=df1, dfd=df2)

print(f"\n--- Resultados da Distribuição F(49, 49) ---")
print(f"P(F <= {f_low}) = {p_cdf_low:.4f} ({p_cdf_low*100:.2f}%)")
print(f"P({f_low} <= F <= {f_high}) = {p_interval:.4f} ({p_interval*100:.2f}%)")
print(f"P(F > {f_high}) = {p_sf_high:.4f} ({p_sf_high*100:.2f}%)")
print(f"P(F >= F_observado = {f_obs:.4f}) = {p_value_f_obs:.4f} ({p_value_f_obs*100:.2f}%)")
print(f"90º Percentil de F = {f_p90:.4f}")
print(f"95º Percentil de F (limiar alfa = 0.05) = {f_p95:.4f}")

# 4. Geração dos Gráficos com Áreas Sombreadas
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
f_axis = np.linspace(0.01, 3.0, 500)
pdf_vals = f.pdf(f_axis, dfn=df1, dfd=df2)

# Gráfico 1: P(F <= 0.70)
axes[0].plot(f_axis, pdf_vals, 'k-', lw=2)
f_fill1 = f_axis[f_axis <= f_low]
axes[0].fill_between(f_fill1, f.pdf(f_fill1, dfn=df1, dfd=df2), color='skyblue', alpha=0.6, label=f'P(F ≤ 0.70) = {p_cdf_low:.4f}')
axes[0].axvline(f_low, color='blue', linestyle='--')
axes[0].set_title('Cauda Inferior (CDF)')
axes[0].set_xlabel('Estatística F (adimensional)')
axes[0].set_ylabel('Densidade f(F)')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Gráfico 2: P(0.70 <= F <= 1.50)
axes[1].plot(f_axis, pdf_vals, 'k-', lw=2)
f_fill2 = f_axis[(f_axis >= f_low) & (f_axis <= f_high)]
axes[1].fill_between(f_fill2, f.pdf(f_fill2, dfn=df1, dfd=df2), color='limegreen', alpha=0.6, label=f'P(0.70 ≤ F ≤ 1.50) = {p_interval:.4f}')
axes[1].axvline(f_low, color='green', linestyle='--')
axes[1].axvline(f_high, color='green', linestyle='--')
axes[1].set_title('Intervalo Central de F')
axes[1].set_xlabel('Estatística F (adimensional)')
axes[1].set_ylabel('Densidade f(F)')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

# Gráfico 3: P(F > 1.50) e F observado
axes[2].plot(f_axis, pdf_vals, 'k-', lw=2)
f_fill3 = f_axis[f_axis >= f_high]
axes[2].fill_between(f_fill3, f.pdf(f_fill3, dfn=df1, dfd=df2), color='salmon', alpha=0.6, label=f'P(F > 1.50) = {p_sf_high:.4f}')
axes[2].axvline(f_obs, color='darkred', linestyle='-', lw=2, label=f'F obs = {f_obs:.4f}')
axes[2].axvline(f_p90, color='orange', linestyle=':', lw=2, label=f'P90 (F = {f_p90:.2f})')
axes[2].set_title('Cauda Superior (SF) e Valor Observado')
axes[2].set_xlabel('Estatística F (adimensional)')
axes[2].set_ylabel('Densidade f(F)')
axes[2].legend()
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

#Interpretação Biológica
##Esta probabilidade refere-se à distribuição da razão de variâncias entre dois grupos sob o modelo nulo, e não à probabilidade de uma planta de Setosa ter sépala maior ou mais larga que uma de Versicolor. As áreas sombreadas abaixo da curva $f(F)$ com $df_1=49$ e $df_2=49$ traduzem visualmente as regiões de aceitação e rejeição para o teste de hipóteses sobre razões de variabilidade biológica.
Logo temos que:
##$10{,}77\%$ é a probabilidade de observar uma razão de variâncias $F \le 0{,}70$ caso as duas populações fossem igualmente variáveis.
##$81{,}26\%$ é a probabilidade de a razão entre duas variâncias amostrais de 50 indivíduos flutuar no intervalo esperado de $0{,}70$ a $1{,}50$.$7{,}97\%$ é a probabilidade de obter $F > 1{,}50$ apenas por flutuação amostral aleatória.
##O $90^\circ$ percentil é $F_{0{,}90} = 1{,}4464$.
##Análise da Amostra Real: A razão das variâncias amostrais calculadas entre Iris setosa ($0{,}1437\text{ cm}^2$) e Iris versicolor ($0{,}0985\text{ cm}^2$) foi $F_{\text{obs}} = 1{,}4592$. A probabilidade de obter $F \ge 1{,}4592$ sob a hipótese de igualdade de variâncias é de $9{,}47\%$ ($p\text{-valor} = 0{,}0947$). Ao nível de significância padrão de $\alpha = 0{,}05$ ($F_{0{,}95} = 1{,}6073$), não rejeitamos a hipótese de igualdade de variâncias.
