```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
# 1. DataFrame com os dados do Exercício 1
dados_ex1 = {
    "Area": [1, 2, 3, 4, 5],
    "Antes": [8, 10, 7, 9, 11],
    "Depois": [13, 14, 11, 15, 14]
}
df_ex1 = pd.DataFrame(dados_ex1)

# Cálculo da diferença
df_ex1["Diferenca"] = df_ex1["Depois"] - df_ex1["Antes"]

# 2. Estatísticas descritivas (Médias e Desvios-Padrão)
media_antes = df_ex1["Antes"].mean()
sd_antes = df_ex1["Antes"].std(ddof=1)

media_depois = df_ex1["Depois"].mean()
sd_depois = df_ex1["Depois"].std(ddof=1)

media_dif = df_ex1["Diferenca"].mean()
sd_dif = df_ex1["Diferenca"].std(ddof=1)

print("--- EXERCÍCIO 1: ESTATÍSTICAS DESCRITIVAS ---")
print(f"Média Antes: {media_antes:.2f} (DP = {sd_antes:.2f})")
print(f"Média Depois: {media_depois:.2f} (DP = {sd_depois:.2f})")
print(f"Diferença das Médias (Depois - Antes): {media_depois - media_antes:.2f}")
print(f"Média das Diferenças: {media_dif:.2f} (DP = {sd_dif:.2f})\n")

# 3. Teste t pareado (scipy.stats.ttest_rel)
# H0: mu_depois - mu_antes = 0
# H1: mu_depois - mu_antes != 0 (ou H1: mu_depois > mu_antes)
res_ex1 = stats.ttest_rel(df_ex1["Depois"], df_ex1["Antes"], alternative="two-sided")

print("--- EXERCÍCIO 1: TESTE T PAREADO ---")
print(f"T = {res_ex1.statistic:.3f}")
print(f"p-valor = {res_ex1.pvalue:.4f}")
print(f"gl = {res_ex1.df}")
print(res_ex1)

alpha = 0.05
decisao_ex1 = "Rejeitar H0" if res_ex1.pvalue <= alpha else "Não rejeitar H0"
print(f"Decisão (alpha=0.05): {decisao_ex1}\n")
print("="*60 + "\n")


# EXERCÍCIO 2: Teste t para Amostras Independentes

# 1.  DataFrame com os dados do Exercício 2
dados_ex2 = {
    "Fragmento": [1, 2, 3, 4, 5],
    "Preservada": [18, 20, 17, 22, 19],
    "Perturbada": [10, 12, 9, 13, 11]
}
df_ex2 = pd.DataFrame(dados_ex2)

preservada = df_ex2["Preservada"]
perturbada = df_ex2["Perturbada"]

# 2. Estatísticas descritivas
media_pres = preservada.mean()
sd_pres = preservada.std(ddof=1)

media_pert = perturbada.mean()
sd_pert = perturbada.std(ddof=1)

print("--- EXERCÍCIO 2: ESTATÍSTICAS DESCRITIVAS ---")
print(f"Média Floresta Preservada: {media_pres:.2f} (DP = {sd_pres:.2f})")
print(f"Média Floresta Perturbada: {media_pert:.2f} (DP = {sd_pert:.2f})\n")

# 3. Teste t para duas médias independentes com variâncias iguais (scipy.stats.ttest_ind)
# H0: mu_preservada = mu_perturbada
# H1: mu_preservada != mu_perturbada
res_ex2 = stats.ttest_ind(preservada, perturbada, equal_var=True, alternative="two-sided")

gl_ex2 = len(preservada) + len(perturbada) - 2

print("--- EXERCÍCIO 2: TESTE T INDEPENDENTE (VARIÂNCIAS IGUAIS) ---")
print(f"T = {res_ex2.statistic:.3f}")
print(f"p-valor = {res_ex2.pvalue:.4f}")
print(f"gl = {gl_ex2}")
print(res_ex2)

decisao_ex2 = "Rejeitar H0" if res_ex2.pvalue <= alpha else "Não rejeitar H0"
print(f"Decisão (alpha=0.05): {decisao_ex2}")
```
