##Exercício 1 — Massa corporal de mamíferos
##A massa corporal de uma espécie de pequeno mamífero apresenta distribuição normal:
##• μ=420 g
##• σ=55 g
##Determine:
##1. A probabilidade de um indivíduo pesar menos de 350 g.
##2. A probabilidade de pesar mais de 500 g.
##3. A probabilidade de pesar entre 400 e 480 g.
##4. Represente graficamente as três probabilidades.

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# Parâmetros da distribuição
mu = 420
sigma = 55

# 1. Cálculo das probabilidades
p1 = stats.norm.cdf(350, loc=mu, scale=sigma)
p2 = 1 - stats.norm.cdf(500, loc=mu, scale=sigma)
p3 = stats.norm.cdf(480, loc=mu, scale=sigma) - stats.norm.cdf(400, loc=mu, scale=sigma)

# Exibição dos resultados no console
print(f"1. P(X < 350 g) = {p1:.4f} ({p1*100:.2f}%)")
print(f"2. P(X > 500 g) = {p2:.4f} ({p2*100:.2f}%)")
print(f"3. P(400 < X < 480 g) = {p3:.4f} ({p3*100:.2f}%)\n")

# 2. Construção dos gráficos
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y = stats.norm.pdf(x, loc=mu, scale=sigma)

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Gráfico 1: P(X < 350 g)
axes[0].plot(x, y, 'b-', lw=2)
x1 = np.linspace(mu - 4*sigma, 350, 500)
axes[0].fill_between(x1, stats.norm.pdf(x1, mu, sigma), color='skyblue', alpha=0.6)
axes[0].axvline(350, color='red', linestyle='--', label='X = 350 g')
axes[0].set_title(f'1. P(X < 350 g) = {p1*100:.2f}%')
axes[0].set_xlabel('Massa corporal (g)')
axes[0].set_ylabel('Densidade de Probabilidade')
axes[0].grid(True, alpha=0.3)
axes[0].legend()

# Gráfico 2: P(X > 500 g)
axes[1].plot(x, y, 'b-', lw=2)
x2 = np.linspace(500, mu + 4*sigma, 500)
axes[1].fill_between(x2, stats.norm.pdf(x2, mu, sigma), color='orange', alpha=0.6)
axes[1].axvline(500, color='red', linestyle='--', label='X = 500 g')
axes[1].set_title(f'2. P(X > 500 g) = {p2*100:.2f}%')
axes[1].set_xlabel('Massa corporal (g)')
axes[1].set_ylabel('Densidade de Probabilidade')
axes[1].grid(True, alpha=0.3)
axes[1].legend()

# Gráfico 3: P(400 < X < 480 g)
axes[2].plot(x, y, 'b-', lw=2)
x3 = np.linspace(400, 480, 500)
axes[2].fill_between(x3, stats.norm.pdf(x3, mu, sigma), color='green', alpha=0.6)
axes[2].axvline(400, color='red', linestyle='--')
axes[2].axvline(480, color='red', linestyle='--', label='Intervalo [400, 480]')
axes[2].set_title(f'3. P(400 < X < 480 g) = {p3*100:.2f}%')
axes[2].set_xlabel('Massa corporal (g)')
axes[2].set_ylabel('Densidade de Probabilidade')
axes[2].grid(True, alpha=0.3)
axes[2].legend()

plt.tight_layout()
plt.show()
```

## 2- $P(X > 500\text{ g})$: $0{,}0729$ ou $7{,}29\%$
## 3- $P(400 < X < 480\text{ g})$: $0{,}5043$ ou $50{,}43\%$

##Exercício 2 — Percentis de tamanho corporal
##O comprimento corporal desses mamíferos segue uma distribuição normal com:
##• μ=150 cm
##• σ=10,5 cm
##Calcule:
##1. O percentil 80.
##2. O valor correspondente a 10% dos indivíduos apresentarem o comprimento superior a esse valor.
##3. Faça um único gráfico indicando os três percentis.
##Interprete biologicamente cada resultado. Por exemplo: qual comprimento separa os 10% maiores indivíduos dos demais?

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# Parâmetros da distribuição
mu = 150
sigma = 10.5

# 1. Cálculo dos percentis (função PPF - Percent Point Function)
p50 = stats.norm.ppf(0.50, loc=mu, scale=sigma) # Mediana (referência)
p80 = stats.norm.ppf(0.80, loc=mu, scale=sigma) # Percentil 80
p90 = stats.norm.ppf(0.90, loc=mu, scale=sigma) # Percentil 90 (10% maiores)

print(f"P50 (Mediana): {p50:.2f} cm")
print(f"P80 (Percentil 80): {p80:.2f} cm")
print(f"P90 (10% maiores): {p90:.2f} cm\n")

# 2. Construção da curva normal e gráfico único
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y = stats.norm.pdf(x, loc=mu, scale=sigma)

plt.figure(figsize=(10, 6))
plt.plot(x, y, 'b-', lw=2, label='Distribuição Normal (μ=150 cm, σ=10.5 cm)')

# Linhas verticais dos percentis
plt.axvline(p50, color='gray', linestyle='--', lw=1.5, label=f'P50 (Mediana) = {p50:.2f} cm')
plt.axvline(p80, color='orange', linestyle='--', lw=2, label=f'P80 = {p80:.2f} cm')
plt.axvline(p90, color='red', linestyle='--', lw=2, label=f'P90 (10% maiores) = {p90:.2f} cm')

# Destaque das áreas sob a curva
plt.fill_between(x, y, where=(x <= p80), color='skyblue', alpha=0.3, label='80% dos indivíduos (≤ P80)')
plt.fill_between(x, y, where=(x >= p90), color='salmon', alpha=0.5, label='10% maiores indivíduos (≥ P90)')

# Formatação do gráfico
plt.title('Distribuição do Comprimento Corporal e Percentis (P50, P80, P90)', fontsize=13)
plt.xlabel('Comprimento corporal (cm)', fontsize=11)
plt.ylabel('Densidade de Probabilidade', fontsize=11)
plt.grid(True, alpha=0.3)
plt.legend(loc='upper left')

plt.tight_layout()
plt.show()
```

##Resultados CalculadosPara uma distribuição normal
##1- $X \sim \mathcal{N}(\mu = 150\text{ cm}, \sigma = 10{,}5\text{ cm} $:Percentil 80 ($P_{80}$): $158{,}84\text{ cm}$10%
##2- Superiores / Percentil 90 ($P_{90}$): $163{,}46\text{ cm}$
##3- Percentil 50 / Mediana ($P_{50}$): $150{,}00\text{ cm}$ (usado como linha central de referência)

##Interpretação biológica
##O percentil 80 indica que 80% dos mamíferos dessa espécie possuem comprimento corporal de até 158,84 cm, ou seja, apenas 20% da população ultrapassa essa média.
##O porcentil de 90, indica que os 10% superiores tem o comprimento maior do que 163,46cm, sendo esse o limite biológico.
##O percentil de 50 (150cm)representa a mediana da populaçãom, separando os indivíduos exatamento no meio, em 50% menores e 50% maiores


##Exercício 3 — Integração
##Uma população de aves apresenta massa corporal normalmente distribuída com:
##• μ=72 g
##• σ=8 g
##Em Python:
##1. Calcule P(X≤60).
##2. Calcule P(X≥85).
##3. Calcule P(65≤X≤80).
##4. Calcule os percentis 90, 95 e 99.
##5. Produza um gráfico da distribuição normal mostrando as três regiões de probabilidade e os três percentis.
##6. Escreva uma interpretação ecológica curta dos resultados.

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# Parâmetros da distribuição
mu = 72
sigma = 8

# 1. Probabilidades
p_le_60 = stats.norm.cdf(60, loc=mu, scale=sigma)
p_ge_85 = 1 - stats.norm.cdf(85, loc=mu, scale=sigma)
p_between_65_80 = stats.norm.cdf(80, loc=mu, scale=sigma) - stats.norm.cdf(65, loc=mu, scale=sigma)

# 2. Percentis
p90 = stats.norm.ppf(0.90, loc=mu, scale=sigma)
p95 = stats.norm.ppf(0.95, loc=mu, scale=sigma)
p99 = stats.norm.ppf(0.99, loc=mu, scale=sigma)

# Exibição dos resultados numéricos
print("--- PROBABILIDADES ---")
print(f"P(X <= 60 g) = {p_le_60:.4f} ({p_le_60*100:.2f}%)")
print(f"P(X >= 85 g) = {p_ge_85:.4f} ({p_ge_85*100:.2f}%)")
print(f"P(65 <= X <= 80 g) = {p_between_65_80:.4f} ({p_between_65_80*100:.2f}%)\n")

print("--- PERCENTIS ---")
print(f"Percentil 90 (P90) = {p90:.2f} g")
print(f"Percentil 95 (P95) = {p95:.2f} g")
print(f"Percentil 99 (P99) = {p99:.2f} g\n")

# 3. Gráficos (Visualização Integrada)
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y = stats.norm.pdf(x, loc=mu, scale=sigma)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Subplot 1: Regiões de Probabilidade
ax1 = axes[0]
ax1.plot(x, y, 'b-', lw=2, label='Distribuição Normal (μ=72, σ=8)')

x_low = np.linspace(mu - 4*sigma, 60, 300)
ax1.fill_between(x_low, stats.norm.pdf(x_low, mu, sigma), color='skyblue', alpha=0.6, label='P(X ≤ 60 g) = 6.68%')

x_mid = np.linspace(65, 80, 300)
ax1.fill_between(x_mid, stats.norm.pdf(x_mid, mu, sigma), color='lightgreen', alpha=0.6, label='P(65 ≤ X ≤ 80 g) = 65.06%')

x_high = np.linspace(85, mu + 4*sigma, 300)
ax1.fill_between(x_high, stats.norm.pdf(x_high, mu, sigma), color='salmon', alpha=0.6, label='P(X ≥ 85 g) = 5.21%')

ax1.set_title('1. Regiões de Probabilidade', fontsize=13)
ax1.set_xlabel('Massa corporal (g)')
ax1.set_ylabel('Densidade de Probabilidade')
ax1.grid(True, alpha=0.3)
ax1.legend(loc='upper right')

# Subplot 2: Percentis
ax2 = axes[1]
ax2.plot(x, y, 'b-', lw=2, label='Distribuição Normal (μ=72, σ=8)')

ax2.axvline(p90, color='orange', linestyle='--', lw=2, label=f'P90 = {p90:.2f} g')
ax2.axvline(p95, color='red', linestyle='--', lw=2, label=f'P95 = {p95:.2f} g')
ax2.axvline(p99, color='darkred', linestyle='--', lw=2, label=f'P99 = {p99:.2f} g')

ax2.set_title('2. Percentis (P90, P95, P99)', fontsize=13)
ax2.set_xlabel('Massa corporal (g)')
ax2.set_ylabel('Densidade de Probabilidade')
ax2.grid(True, alpha=0.3)
ax2.legend(loc='upper left')

plt.tight_layout()
plt.show()
```

##Para uma população de aves com massa corporal distribuída normalmente $X \sim \mathcal{N}(\mu = 72\text{ g}, \sigma = 8\text{ g})$:
##1. $P(X \le 60\text{ g})$: $0{,}0668$ ou $6{,}68\%$
##2. $P(X \ge 85\text{ g})$: $0{,}0521$ ou $5{,}21\%$
##3. $P(65 \le X \le 80\text{ g})$: $0{,}6506$ ou $65{,}06\%$
##4. Percentis:
##$P_{90}$: $82{,}25\text{ g}$
##$P_{95}$: $85{,}16\text{ g}$
##$P_{99}$: $90{,}61\text{ g}$

##Interpretação Ecológica

##    A maioria fica no peso ideal (65% das aves): A grande maioria pesa entre 65g e 80g. Para a sobrevivência da espécie, esse peso médio é o mais vantajoso, pois garante reservas de energia sem prejudicar a agilidade do voo.

##Aves muito leves (6,7% abaixo de 60g): São poucas as aves bem magras. Elas enfrentam maior perigo em épocas de frio ou falta de alimento por não terem gordura estocada.

##Aves muito pesadas (5,2% acima de 85g): Poucas aves chegam a esse peso. A gordura extra serve para disputar território ou aguentar longas migrações, mas torna o voo mais pesado e dificulta a fuga de predadores.

##Os limites de tamanho (Percentis): Apenas 10% das aves pesam mais de 82,25g (P90​) e é raríssimo encontrar uma com mais de 90,61g (P99​, apenas 1% do grupo). Esses valores servem como referência para saber se uma ave capturada está muito acima da média.
