```python
import pandas as pd

# Replace 'your_file.xlsx' with the path to your XLSX file
# For demonstration, let's assume a file named 'example.xlsx' is uploaded or available.
# If you upload a file, it will typically be in the root directory, e.g., 'example.xlsx'

try:
    df = pd.read_excel('dados.xlsx')
    print('Successfully loaded dados.xlsx:')
    display(df.head())
except FileNotFoundError:
    print("Error: 'dados.xlsx' not found. Please upload your XLSX file or provide the correct path.")
except Exception as e:
    print(f"An error occurred: {e}")
```
