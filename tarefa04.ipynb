#**Quartil**
##Os quartis são utulizados para dividir um conjunto de dados em quatro partes iguais (25%), utilizando a sigla ***Q***.
##O primeiro quartil, Q₁, corresponde a 25% aos valores que são menores ou iguais em relação aos outros 75%. Tendo como exemplo o nosso banco de dados, vemos que 25%, que corresponde dos 9,25 das amostras, possuem valor igual ou inferior a 21,0 (mm) de largura. Podendo ser calculado pela equação: $$ Q₁=\frac{(n+1)}{4} $$
##Q₂ corresponde a mediana, ou seja, 50% dos valores são menores e os outros 50% são maiores, sendo exatamente a metade do total. Seguindo o exemplo, exatamente a médias das amostras correspondem a uma largura de 31,20mm, logo sabemos que 50% é igual ou inferior a essa medida.
##Q₃ corresponde ao último 1/4 do valor total, ou seja, 75% dos valores são menores ou inferiores, logo 25% são maiores ou iguais ao terceiro quartil. No mesmo exemplo, o limite de 75% é de 50,50mm, logo todos os valores iguais ou inferios a essa medida são incluídas nesse quartil. Podemos calcular pela equação: $$ Q₃=\frac{3(n+1)}{4} $$



#**Mediana**
##Corresponde ao valor do meio de uma disposição de dados ordenados classificados do menor para o maior, onde metade dos valores são inferiores ou iguais e a outra metade são superiores ou iguais. Para calcular a mediana, primeiro é necessário ordenar o valor, sempre do menor para o maior, onde é utilizado a seguinte equação para calcular o exato valor que corresponde a mediana: $$ Q₁=\frac{(n+1)}{2} $$
##Utilizamos também duas regras básicas, sendo elas:
###***regra 1:*** Se o conjunto de dados for um número impar, a mediana irá corresponder a medição associada ao valor que se encontra no meio (como é o caso da tabela de dados utilizada)
###***regra 2:*** Se o conjunto de dados for um número impar, a mediana irá corresponder a *média* associada ao valor que se encontra no meio
##Utilizando os dados da tabela como exemplo, ordenamos a sequência na qual queremos achar a mediana, como feito em comandos posteriores utilizando o código: sort_index(ascending=True).
## onde foi ordenado os números de forma crescente da largura dos frutos, ex: $$ \frac{(4,2 - 4,5 - 5,3 ... 129,2 - 150,0 - 175,0)}{1 - 2 - 3 ... 35 - 36 - 37 } $$
##Como os dados apresentam um número ímpar na ordem de classificação (37), pegaremos exatamente a linha que está no meio da ordem crescente, que corresponde a linha 19, onde o valor é de 31,20mm, sendo essa a nossa mediana, então todos os demais valores, metade serão superior e a outra metade inferior a essa medida.

#**Amplitude interquartílica**
##Também é conhecida como dispersão média, corresponde a diferença entre o terceiro e o primeiro quartil de um conjunto de dados: Amplitude interquartil = Q₃ - Q₁. Essa amplitudo é fundamental para medir a dispersão dos dados sem extremos, pois não leva em consideração os valores de 25% maiores e menores e evita também a distorção de dados, tendo uma maior distribuição quando há valores muito baixos ou muito altos, não recebendo influência exagerada desses valores.  
##No nosso banco de dados, sabemos que o valor de Q₃ é 50,50mm e o de Q₁ é de 21,0mm, colocando na equação de amplitude interquaril, vemos que:
$$ IQR=\frac{(50,5)}{21}=29,5mm $$
##Sendo esse o valor da metade central da largura dos frutos.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel('/content/tarefa04.xlsx')
```

```python
df
```

```python
#quartis
largura_do_fruto= df["Largura do fruto(mm)"].value_counts().sort_index(ascending=True)
```

```python
display(largura_do_fruto)
```

```python
dados= df['Largura do fruto(mm)'].dropna()

q1= dados.quantile(0.25)
q2= dados.quantile(0.5)
q3= dados.quantile(0.75)
print(f"Q1: {q1:.2f} | Q2 (Mediana): {q2:.2f} | Q3: {q3:.2f}")
```

#**Interpretação dos quartis**
##q1: 25% das amostras analisadas (total=37, 25%=9,25) possuem valor igual ou inferior a 21,0 (mm) de largura
##q2: o quartil 2 indica a mediana dos dados, que no caso, como a quantidade de amostras analisadas é igual a um número ímpar (37), se trata do elemento central/a 19°posição (31,20). Isso significa que metade dos valores obtidos no estudo é igual ou inferior à 31,20 (mm) enquanto a outra metade é superior a essa medida de largura.
##q3: 75% dos dados possuem valor igual ou inferior a 50,50 (mm), enquanto apenas 25% dos dados são valores superiores a esse.

```python
#Boxplot
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8,5))

ax=sns.boxplot(x=df['Largura do fruto(mm)'], color='#008000', width=0.4, fliersize=6)

plt.title('Distribuição da Largura dos Frutos consumidos por Cachorro do mato', fontsize=13, fontweight='bold', pad=15)
plt.xlabel('Largura do Fruto (mm)', fontsize=11, labelpad=10)
plt.tight_layout()
plt.show()
```

#**Interpretação boxpot**
##No gráfico, a caixa verde apresenta onde estão os valores centrais (de 21,00 mm a 50,50 mm), localizados entre o q1 e q3. Isso corresponde à Amplitude Interquartil (IQR) e indica a dispersão de 50% dos dados centrais da amostra.

##O IQR é calculado a partir da fórmula: IQR= q3-q1

##No caso dos dados de largura o IQR seria igual a 29,50. Desse modo, os limites superiores e inferiores seriam calculados a partir da fórmula:
###Limite inferior= q1 - (1,5*29,50) = -23,25
###Limite superior= q3 + (1,5*29,50) = 94,75
##No entanto, esses limites são teóricos, não podendo ser aplicado um limite negativo para a largura de sementes, que não pode ser um dado negativo na realidade. Desse modo, a haste desenhada no boxplot segue até o menor dado real da tabela que seja maior ou igual a esse limite.

##Os limites inferior e superior são fundamentais para desenhar observações atípicas. Quaisquer números que estejam fora dos limites calculados são denominados de outliers. Esse é o caso dos dados (95,0mm;115,0mm; 129,2mm; 150,0mm e 175,0mm) que se encontram como pontos à direita do limite superior.


#**Como determinamos se há observações atípicas em uma base de dados?**
##A determinação de observações atípicas (outliers) ocorre  por meio do cálculo dos limites inferior e superior do boxplot. Eles definem a fronteira aceitável de distribuição.  permitindo observar os dados "fora das linhas", ou seja, que não se encontram entre os limites estabelecidos. Essa etapa é fundamental no tratamento de dados científicos, pois possibilita identificar padrões atípicps de interesse biológico, como anomalias ou erros metodológicos.
