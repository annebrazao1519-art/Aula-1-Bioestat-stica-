#Pergunta 2
##Gráfico de dispersão:

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_excel('tarefa06.xlsx')

X = df['Temperatura Ambiental (°C)'].values
Y = df['Taxa Metabólica em Repouso (mL O₂/g/h)'].values

beta_1, beta_0 = np.polyfit(X, Y, 1)

print(f"Intercepto (beta_0): {beta_0:.2f}")
print(f"Coeficiente Angular (beta_1): {beta_1:.2f}")
print(f"Equação da Reta: Y = {beta_0:.2f} + {beta_1:.2f} * X")

plt.figure(figsize=(8, 5))

plt.scatter(X, Y, color='darkblue', s=70, label='Dados Observados')

plt.plot(X, beta_0 + beta_1 * X, color='red', linestyle='--', linewidth=2,
         label=f'Reta Ajustada ($\hat{{y}} = {beta_0:.2f} + {beta_1:.2f}x$)')

plt.title('Regressão Linear Simples: Temperatura vs Taxa Metabólica', fontsize=13, fontweight='bold', pad=15)
plt.xlabel('Temperatura Ambiental (°C)', fontsize=11)
plt.ylabel('Taxa Metabólica em Repouso (mL $O_2$/g/h)', fontsize=11)
plt.legend(loc='upper left', fontsize=10)
plt.grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```

#Pergunta 3

##Uma regressão linear simples está relacionada a previsão de "tendencias". É basicamente a analise de dados empiricos, ou seja, que foram coletados através de observação. Ao inves de só tentar adivinhar um padrão através desses dados, a regressão linear simples consegue prever se estão relacionados, a intensidade e até previsões futuras.

##Nela exitem duas variaveis: X (variável independente, no nosso caso é representado pela temperatura) e Y (variável dependente, taxa metabólica do animal). O objetivo desse metodo é traçar uma linha para simular como Y muda em diferentes valores de X, expressa pela equação:

$$Y = \beta_0 + \beta_1 X + \epsilon$$

##O método mais utilizado para calcular isso é a Métrica dos Mínimos Quadrados Ordinários (OLS), onde ele pega os dados reais observados na amostra para calcular a distancia (erros) dos pontos e da reta, encontrando uma estimativa para os parametros da equação de modelo populacional, calculado por:

$$\hat{Y} = \hat{\beta}_0 + \hat{\beta}_1 X$$




#Pergunta 4
##Assim como visto na resposta anterior, a regressão linear simples é calculado atrês do código a serguir, sendo uma soma de cálculos que obtém o menor número de erros para formar uma linha reta hipotética, chegando a conclusão que o coeficiente angular é: $$\beta_1 = \frac{\sum (X - \bar{X})(Y - \bar{Y})}{\sum (X - \bar{X})^2} = \mathbf{62{,}67}$$
##que indica a taxa de variação metabólica por grau Celsius.
##E o Intercepto resulta no valor de :$$\beta_0 = \bar{Y} - \beta_1 \bar{X} = \mathbf{-1317{,}75}$$
##Ponto exato onde a reta cruza o eixo Y (Igualando o x a 0)

```python
import pandas as pd
import numpy as np

# 1. Carregar os dados
df = pd.read_excel('tarefa06.xlsx')

X = df['Temperatura Ambiental (°C)']
Y = df['Taxa Metabólica em Repouso (mL O₂/g/h)']

# 2. Estimar os coeficientes da reta: Y = beta_0 + beta_1 * X
# O parâmetro '1' indica uma regressão polinomial de 1º grau (linha reta)
beta_1, beta_0 = np.polyfit(X, Y, 1)

# 3. Exibir os resultados formatados
print("=== ESTIMAÇÃO DOS COEFICIENTES ===")
print(f"Coeficiente Angular (beta_1 / Inclinação): {beta_1:.4f}")
print(f"Intercepto (beta_0 / Constante): {beta_0:.4f}")
print(f"\nEquação da Reta Ajustada: Y = {beta_0:.2f} + {beta_1:.2f}")
```

#Pergunta 5
##Interpretação dos dados
##Coenficiênte angular: representa a taxa de variação biológica. Como o programa indicou um coeficiente angular de: $\beta_1 = +62,67$. Interpretamos que para cada $1^\circ\text{C}$ no aumento da temperatura ambiente, o organismo tem um aumento da taxa metabólica em média de $62{,}67\text{ mL O}_2/\text{g/h}$.
##Em termos biológicos, esse aumento indica a sensibilidade metabólica do animal em relação a temperatura. Como o estudo foi relizado em animais ectotérmico, onde a temperatura corporal é influênciada diretamente pela temperatura ambiente, podemos concluir com a interpretação dos dados analisados que o aumento da temperatura, desencadeia no aumento das reações metabólicas, exigindo um maior consumo de oxigênio por hora ($\text{ mL O}_2/\text{g/h}$).

##Intercepto: é o valor estimado para a taxa metabólica ($Y$) quando a temperatura ambiental ($X$) é exatamente igual a $0^\circ\text{C}$.Ou seja, indica o consumo de oxigênio a $0^\circ\text{C}$
##Portanto, podemos concluir que o intercepto não tem valor nenhum biológico, pois se analisarmos o valor de $\beta_0 = \bar{Y} - \beta_1 \bar{X} = \mathbf{-1317{,}75}$, concluimos rapidamente que um organismos não poderia viver em um ambiente onde o oxigênio é negativo.
##No entando, matematicamente, esse valor é fundamental para que a reta passe pelo centro dos dados observados, sem o intercpto, a reta passaria pela origem, que distorceria por completo os dados.





#Pergunta 7
## A correlação linear de Pearson representa, a partir de um coeficiente r, o quanto duas variáveis se alinham ao longo de uma reta, ou seja, a correlação entre elas. Essa correlação pode ser direta, inversa ou inexistente (nula). O r varia de 1 a -1 e indica a direção e a força dessa relação. O -1 indica uma relação negativa, ou seja,que elas apresentam movimentação oposta, se uma cresce a outra decresce. O extremo positivo 1 indica que as variáveis se movimentam na mesma direção, crescendo ou decrescendo juntas. O 0, por sua vez, indica que não há relação linear entre elas.

```python
import scipy.stats as stats
r, p_valor = stats.pearsonr(
    df['Temperatura Ambiental (°C)'],
    df['Taxa Metabólica em Repouso (mL O₂/g/h)']
    )

print(f"Coeficiente de Correlação de Pearson (r): {r:.4f}")
print(f"P-valor: {p_valor:.4e}")
```

Desse modo, é possível afirmar que ambas andam da mesma direção, quando a temperatura ambiental diminui a taxa metabólica também decai.

#Pergunta 10

##1. Modelo de Regressão Linear Simples


$$Y = \beta_0 + \beta_1 X + \epsilon$$

##2. Estimadores OLS (Mínimos Quadrados Ordinários)


* **Coeficiente Angular ($\hat{\beta}_1$):**
$$\hat{\beta}_1 = \frac{\sum_{i=1}^{n} (X_i - \bar{X})(Y_i - \bar{Y})}{\sum_{i=1}^{n} (X_i - \bar{X})^2}$$

* **Intercepto ($\hat{\beta}_0$):**
$$\hat{\beta}_0 = \bar{Y} - \hat{\beta}_1 \bar{X}$$

##3. Coeficiente de Correlação de Pearson ($r$)

$$r = \frac{\sum_{i=1}^{n} (X_i - \bar{X})(Y_i - \bar{Y})}{\sqrt{\sum_{i=1}^{n} (X_i - \bar{X})^2 \cdot \sum_{i=1}^{n} (Y_i - \bar{Y})^2}}$$

A Correlaçaõ de Pearson permite avaliar a força e a direção da associação entre as variáveis, ajudando a avaliar se os dados que você observou estão relacionados ou se o padrão é fruto do acaso.
